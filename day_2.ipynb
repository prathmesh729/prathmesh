```python
class Area:
    def __init__(self, l, b):
        self.length = l
        self.breadth = b

    def returnArea(self):
        area = self.length * self.breadth
        return area

# Taking length and breadth as input from the keyboard
length = int(input("Enter length: "))
breadth = int(input("Enter breadth: "))

# Creating an object of the Area class
obj = Area(length, breadth)

# Calling returnArea() and printing the result
print("The area of the rectangle is:", obj.returnArea())
```

```python
from abc import ABC, abstractmethod

class Test(ABC):
    @abstractmethod
    def m1(self):
        pass

    @abstractmethod
    def m2(self):
        pass

class MyPython(Test):
    def m1(self):
        print("m1 implementation")

    def m2(self):
        print("m2 implementation")

# Creating an object of the child class and calling the methods
m = MyPython()
m.m1()
m.m2()
```

```python
data = [[[1, 2], [3, 4]], [[5, 6], [7, 8]]]

def fun(m):
    v = m[0][0]

    for row in m:
        for element in row:
            if v < element:
                v = element

    return v

print(fun(data[0]))
```
